```python
# 학생 풀이 ^ 1트
for idx,i in enumerate(df[3]):
    if not isinstance(i, (int, np.int64)) or isinstance(i, (float,np.float64)):
        # isinstance() : 어떤 객체(object)가 특정 클래스(classinfo)의 인스턴스인지 확인할 때 사용
        print(f'열 3 에서 {idx}번째 값이 int64가 아닙니다.')
        print(f'값 : {i}, 타입 : {type(i)}')
# 모든 데이터의 값이 문자열이라서 열3의 전체 데이터가 출력되었다.
```

```python
for idx, value in enumerate(df[3]):
    if not isinstance(value, (int, float, str, np.int64, np.float64)):
        print(f"열 3의 {idx}번째 값이 정수, 실수, 문자열이 아닙니다.")
        print(f"값: {value}, 타입: {type(value)}")
# Pandas의 object 컬럼에서 NaN 값(np.nan)은 float 타입으로 취급됩니다.
# Boolean 값(True/False)은 bool 타입이지만, 현재 데이터에는 없는 경우라 출력안된다.

# 이렇게 하면서 타입들을 하나씩 추가하면서 알아가기보단 기대된 정상적인 데이터의 타입을 파악해두고
# 실무에서는 밝혀지지않고제시될 데이터 결측치의 정체를 알기위해서
# 결측치의 타입을  추론해 알아가는 과정에서나 사용할듯하다.
# 아니면 직접 보면서 결측치의 데이터가 뭘로 채워져있는지 확인하는 방법도 직관적일듯 하다.
```

```python
# 결측치로 저리해야될 값의 타입은 문자-문자열이고, 열 3 에 있다.
# 수로 변환할수없을 텐데?
```

```python
# 학생 풀이 ^ 2트 gpt
# 열 3의 데이터에서 숫자 형태가 아닌 값을 찾는 함수
# columns = [3]
def find_non_numeric(df, columns):
    for col in columns: # 결측치가 예상이되는 컬럼 리스트에서 컬럼을 순환, 모든 결측치가 있는 컬럼을 순환, 연산한다.
        for idx, value in enumerate(df[col]): # 해당하는 컬럼의 각행의 인덱스와 값을 순환하고 조회한다.
            try:
                # 문자열인 경우, 공백 제거 후 숫자로 변환 시도
                if isinstance(value, str):
                    value = value.strip()
                pd.to_numeric(value)
            except Exception:
                print(f"열 {col}의 {idx}번째 값이 숫자가 아닙니다.")
                print(f"값: {df[col][idx]}, 타입: {type(df[col][idx])}")

# to_numeric() : 문자열 -> 숫자형 변환하는 함수

find_non_numeric(df,[3])
# https://chatgpt.com/share/68dcb355-ed78-8012-a253-2ea3d903779c

# 임의의 데이터값에서 object 타입으로 변환시킨 결측치 여야할 값을 찾는 예제 풀어보기
```

```python
# 구간을 직접 지정  pd.cut  Domain 지식이 있을때... 
# 마력 low/ medium /hight
import numpy as np
cnt,bins = np.histogram(df[3],bins=3)
print(cnt, bins)
labels = ['저출력','보통','고출력']
df['출력'] = pd.cut(df[3],
    bins = bins,
    labels = labels
)
df.head()
```

```python
# pd.qcut  분위수기반으로 동일한 개수로 분할
# 각 구간별 샘플개수가 비슷
# 구간폭이 일정하지 않음
labels = ['저출력','보통','고출력']
df['출력q'] = pd.qcut(df[3],q=3,labels=labels)
df.head()
```

```python
# 열 이름을 지정
df.columns = ['mpg','cylinders','displacement','horsepower','weight',
              'acceleration','model year','origin','name'] 

# horsepower 열의 누락 데이터('?') 삭제하고 실수형으로 변환
df['horsepower'] = df['horsepower'].replace('?', np.nan)      # '?'을 np.nan으로 변경
df = df.dropna(subset=['horsepower'], axis=0)                 # 누락데이터 행을 삭제
df['horsepower'] = df['horsepower'].astype('float')           # 문자열을 실수형으로 변환

# np.histogram 함수로 3개의 bin으로 나누는 경계 값의 리스트 구하기
count, bin_dividers = np.histogram(df['horsepower'], bins=3)
print(bin_dividers) 
```

```python
# 3개의 bin에 이름 지정
bin_names = ['저출력', '보통출력', '고출력']

# pd.cut 함수로 각 데이터를 3개의 bin에 할당
df['hp_bin'] = pd.cut(x=df['horsepower'],     # 데이터 배열
                      bins=bin_dividers,      # 경계 값 리스트
                      labels=bin_names,       # bin 이름
                      include_lowest=True)    # 첫 경계값 포함 

# horsepower 열, hp_bin 열의 첫 15행을 출력
df[['horsepower', 'hp_bin']].head(15)
```
