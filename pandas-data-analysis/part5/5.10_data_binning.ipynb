```python
# 라이브러리 불러오기
import pandas as pd
import numpy as np

# read_csv() 함수로 df 생성
df = pd.read_csv('./data/auto-mpg.csv', header=None)
df.head()
```

```python
df.info()
# 3번, 8번 항목에서 dtype이 object 타입
# 다른 타입이 섞여 있어서 object 타입으로 변환이 되어있는 것으로 예상하고,
# 안에 있는 다른 타입을 가진 항목을 찾고자 한다.
```

```python
# 학생 풀이 ^
for idx,i in enumerate(df[3]):
    if not isinstance(i, (int, np.int64)):
        # isinstance() : 어떤 객체(object)가 특정 클래스(classinfo)의 인스턴스인지 확인할 때 사용
        print(f'열 3 에서 {idx}번째 값이 int64가 아닙니다.')
        print(f'값 : {i}, 타입 : {type(i)}')
```

```python
# 강사 풀이
for idx,data in enumerate(df[3]):
    try:
        float(data.strip())
    except Exception as e:
        print(df[3][idx], idx)
        print(e)
```

```python
df[3] = df[3].replace('?',pd.NA)
```

```python
import missingno as msno
msno.matrix(df)
# 데이터가 많은 겨우 이런 시각화 하는 라이브러리는 시간이 걸릴수 있다.
# 10만 행 × 10개 열 정도면 일반 시각화 라이브러리에서 버벅임.
# 1,000만 행 이상은 특별한 방법(샘플링/빅데이터 시각화 전용 툴) 없이는 힘듦.
```

```python
df = df.dropna() # ? 부분에 na를 채우고 dropna로 drop하기
# df[3].hist()
# 구간을 나눌 필요성이 있어 보인다.
df[3] = df[3].astype('float') # 구간을 나눈 명령어? ^ 일단은 아래 컬럼 항목이 정리 되었다.
                              # 실수형으로 변환한다고 해서 구간이 나누어지는 건 아니고 도와준다는 느낌에 가까움
df[3].hist(bins=3) # 이게 실제로 구간을 나누는 명령어

# astype('float') → 히스토그램이 구간을 세밀하게 나눌 수 있도록 데이터를 연속형으로 만들어주는 전처리
# hist(bins=3) → 실제로 데이터를 3개의 구간으로 나누어 그래프를 그림
```

```python
# 구간을 직접 지정  pd.cut  Domain 지식이 있을때... 
# 마력 low/ medium /hight
import numpy as np
cnt,bins = np.histogram(df[3],bins=3)
print(cnt, bins)
labels = ['저출력','보통','고출력']
df['출력'] = pd.cut(df[3],
    bins = bins,
    labels = labels
)
df.head()
```

```python
# pd.qcut  분위수기반으로 동일한 개수로 분할
# 각 구간별 샘플개수가 비슷
# 구간폭이 일정하지 않음
labels = ['저출력','보통','고출력']
df['출력q'] = pd.qcut(df[3],q=3,labels=labels)
df.head()
```

```python
# 열 이름을 지정
df.columns = ['mpg','cylinders','displacement','horsepower','weight',
              'acceleration','model year','origin','name'] 

# horsepower 열의 누락 데이터('?') 삭제하고 실수형으로 변환
df['horsepower'] = df['horsepower'].replace('?', np.nan)      # '?'을 np.nan으로 변경
df = df.dropna(subset=['horsepower'], axis=0)                 # 누락데이터 행을 삭제
df['horsepower'] = df['horsepower'].astype('float')           # 문자열을 실수형으로 변환

# np.histogram 함수로 3개의 bin으로 나누는 경계 값의 리스트 구하기
count, bin_dividers = np.histogram(df['horsepower'], bins=3)
print(bin_dividers) 
```

```python
# 3개의 bin에 이름 지정
bin_names = ['저출력', '보통출력', '고출력']

# pd.cut 함수로 각 데이터를 3개의 bin에 할당
df['hp_bin'] = pd.cut(x=df['horsepower'],     # 데이터 배열
                      bins=bin_dividers,      # 경계 값 리스트
                      labels=bin_names,       # bin 이름
                      include_lowest=True)    # 첫 경계값 포함 

# horsepower 열, hp_bin 열의 첫 15행을 출력
df[['horsepower', 'hp_bin']].head(15)
```
