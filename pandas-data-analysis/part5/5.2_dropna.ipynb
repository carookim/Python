```python
# 라이브러리 불러오기
import seaborn as sns

# titanic 데이터셋 가져오기
df = sns.load_dataset('titanic')

# isnull() 메서드를 합계로 집계하여 각 열의 NaN 개수 계산하기
df.isnull().sum(axis=0)
```

```python
df.iloc[:,:2]
print(df.iloc[:,:2].sum()) # 열방향으로 더해진다.
print('*'*60)
print(df.iloc[:,:2].sum(axis=1)) # 행방향으로 더해진다.
```

```python
# NaN가 아닌 값이 500개 이상인 열은 남긴다. - deck 열(891개 중 688개의 NaN 값)
df_thresh = df.dropna(axis=1, thresh=500)  
print(df_thresh.columns) # 남은 열 종목 확인
```

```python
df.info()
# age, deck 에서 결측치가 확인된다.
```

```python
# age 열에 나이 데이터가 한개라도 없는 모든 행을 삭제 - age 열(891개 중 177개의 NaN 값)
df_age = df.dropna(subset=['age'], how='any', axis=0) 
print(len(df_age))
```

```python
# 데이터 프레임에서 삭제가 일어나면 인덱스가 시퀀스 하지않는다. - 값이 재정렬이 되지 않는다.
df_age.reset_index(drop=True)
df.info()
```

```python
# age 열, deck 열 양쪽 모두 데이터가 없는 행을 삭제 
df_age_deck = df.dropna(subset=['age', 'deck'], how='all', axis=0)  
print(len(df_age_deck))
```
