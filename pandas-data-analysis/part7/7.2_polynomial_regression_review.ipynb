##### 선형회귀 분석 - (2) 다항회귀 분석

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# 1️⃣ 예시 데이터 생성 (훈련 데이터)
X_train = np.array([[1], [2], [3], [4], [5]])
y_train = np.array([3, 7, 13, 21, 31])  # y = x^2 + x + 1 형태

# 2️⃣ PolynomialFeatures로 변환 (2차항)
poly = PolynomialFeatures(degree=2, include_bias=False)
X_train_poly = poly.fit_transform(X_train)

# 3️⃣ 변환된 데이터프레임 조회
poly_feature_names = poly.get_feature_names_out(['x'])
df_train_poly = pd.DataFrame(X_train_poly, columns=poly_feature_names)
print("📘 [PolynomialFeatures로 변환된 훈련 데이터]\n")
print(df_train_poly, "\n")

# 4️⃣ 선형회귀 모델 학습
model = LinearRegression()
model.fit(X_train_poly, y_train)

# 5️⃣ 계수와 절편 출력
print("📊 [회귀 결과]")
print("컬럼 이름:", poly_feature_names)
print("회귀 계수:", model.coef_)
print("절편:", model.intercept_)

# 6️⃣ 방정식 형태로 보기 쉽게 표현
equation = "y = " + " + ".join(
    [f"{coef:.3f}*{name}" for coef, name in zip(model.coef_, poly_feature_names)]
) + f" + {model.intercept_:.3f}"
print("\n🧮 회귀식:", equation)
```
