```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
wine_quality = fetch_ucirepo(id=186)

# data (as pandas dataframes)
X = wine_quality.data.features
y = wine_quality.data.targets

# y의 값이 6이상이면 1 그렇지 않으면 0
y2 = y.quality.apply(lambda x: 1 if x >= 6 else 0)
```

```python
# y의 값이 6이상이면 1 그렇지 않으면 0
y2 = y.quality.apply(lambda x: 1 if x >= 6 else 0)
```

```python

# 데이터 분할
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y2, test_size=0.2, random_state=42,stratify=y2)
```

```python

# 랜덤포레스트 모델 학습
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
     

```

```python

# 결정트리 모델 학습
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

```

```python

# rf_model 와 dt_model 예측값
rf_pred = rf_model.predict(X_test)
dt_pred = dt_model.predict(X_test)
# 각 예측값을 이용한 classifier report
print("Random Forest Classifier Report:")
print(classification_report(y_test, rf_pred))
print("\nDecision Tree Classifier Report:")
print(classification_report(y_test, dt_pred))
     
```

```python


# 클래스 불균형을 처리, 0:477 1: 823을 upscaling  SMOTE EEN
from imblearn.over_sampling import SMOTE
from imblearn.combine import SMOTEENN
smote_enn = SMOTEENN(random_state=42)
print(y_train.value_counts())
X_train_resampled2, y_train_resampled2 = smote_enn.fit_resample(X_train, y_train)
print(y_train_resampled2.value_counts())
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train_resampled2, y_train_resampled2)
rf_model_predict = rf_model.predict(X_test)
print(classification_report(y_test, rf_model_predict))
```
