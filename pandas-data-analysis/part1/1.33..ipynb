```python
import pandas as pd
import seaborn as sns
```

```python
mpg = sns.load_dataset('mpg')
mpg.head(3)
```

```python
# groupby - cylindees 별 마력의 평균
            # name 별
            # name이 공백을 기준으로 제조사[0] - 모델
            # 제조사 컬럼을 만들고 제조사를 추가
# iloc, loc 원하는 row col 데이터 추출 - mpg 15이상인 모델
# index를 재구성 - origin
# iloc, loc 를 이용해서 데이터를 row, col다양하게 출력
```

```python
# groupby - cylindees 별 마력의 평균
            # name 별
            # name이 공백을 기준으로 제조사[0] - 모델
            # 제조사 컬럼을 만들고 제조사를 추가
# tips.groupby('day',observed=True)['total_bill'].agg(['sum','mean','max'])
cyl_hp_mean = mpg.groupby('cylinders')['horsepower'].mean() # 실린더별 마력의 평균
mpg['manufacturer'] = mpg['name'].str.split(' ').str[0] # 제조사 분류 및 컬럼 추가
manu_hp_mean = mpg.groupby('manufacturer',observed=True)['horsepower'].mean() # 제조사별 마력 평균
manu_hp_mean
```

```python
name_model_year = mpg.groupby('name')['model_year'].max()
name_model_year
```

```python
mpg['maker'] = [i[0] for i in mpg['name'].str.split().values]
mpg['name'].apply(lambda x: x.split()[0]) # 이름을 람다식으로 첫번째 공백을 기준으로 분리한다.
```
