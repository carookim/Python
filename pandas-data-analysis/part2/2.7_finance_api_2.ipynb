```python
# 라이브러리 가져오기
import requests
import pandas as pd
import os
import dotenv
```

```python
dotenv.load_dotenv()
key = os.getenv('OPEN_DART_KEY')
```

```python
url = 'https://opendart.fss.or.kr/api/list.json'

params = {
    'crtfc_key' : key
}

response = requests.get(url,params=params)

if response.status_code == 200: # 통신성공
    print(response.json())
else:
    print('통신오류')
```
