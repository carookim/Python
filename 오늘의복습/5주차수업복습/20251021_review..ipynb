```
로지스틱 회귀
    시그모이드 함수
    비용함수

로지스틱 회귀 :
    이름에 회귀(regression) 가 들어가지만, 사실은 분류(classification) 모델이야.
    “이 데이터가 1일까 0일까” 같은 이진 분류 문제를 해결한다.

시그모이드 함수 :
    입력값(선형결합 결과)을 0~1 사이의 확률로 바꾼다.
    입력이 크면 1에 가까워지고,
    입력이 작으면 0에 가까워진다.

비용함수 : 
    모델예측과 실제 값을 비교하여 모델을 평가한다.
    비용함수(Cost Function) 이해하기
    비용 = 모델 예측과 실제값의 차이 → 모델이 틀릴수록 비용이 커짐

    예시:
    1. 실제값 y = 1, 모델 예측 0.8 → 약간 틀림 → 비용 작음
    2. 실제값 y = 0, 모델 예측 0.9 → 많이 틀림 → 비용 큼

    즉, 비용은 모델이 잘못 예측한 정도를 숫자로 나타낸 것이라고 생각하면 됩니다.

    2️⃣ 0과 1로 부여하는 방법 (분류 문제)
    - 로지스틱 회귀 같은 이진 분류에서 실제값은 y = 0 또는 y = 1
    - 모델은 ŷ ∈ (0, 1)로 확률을 예측

    비용함수 예시: 로그 손실(log loss, cross-entropy loss)

    비용 = −(y * log(ŷ) + (1 − y) * log(1 − ŷ))

    - y = 1일 때: 비용 = −log(ŷ)
    - y = 0일 때: 비용 = −log(1 − ŷ)

    → 정답에 가까울수록 비용은 작아지고, 정답에서 멀수록 비용이 커짐

    3️⃣ 패널티를 부여하는 방식
    - 정답을 맞추면 패널티(비용) = 0
    - 정답에서 멀어질수록 패널티가 커짐

    예시:
    - 실제값 1, 예측값 0.9 → 비용 ≈ 0.105
    - 실제값 1, 예측값 0.1 → 비용 ≈ 2.3

    → 모델은 학습하면서 비용을 최소화하도록 파라미터를 조정
```
