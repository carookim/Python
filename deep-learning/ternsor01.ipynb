```python
# 기본 연산
import torch

# 빈 텐서 생성
x = torch.empty(3,4) # 3개씩 4개의 컬럼을 가진 빈텐서 생성
x
```

```python
# 랜덤 텐서
x = torch.rand(3,4)
x
```

```python
# 0으로 텐서 초기화
torch.zero(3,4,dtype=torch.long) # .zero() : 텐서 내부를 0으로 채운다.
                                 # dtpye = torch.long : torch.long = torch.int64
```

```python
# 1초 채워진 텐서
torch.ones(3,4,dtype=torch.long) # .one() : 텐서 내부를 1로 채운다.
```

```python
import numpy as np
print(torch.tensor([3.5,3,2.1]))
print(np.array([3.5,3,2.1]))
```

```python
# 기존의 텐서를 기반으로 새로운 텐서 생성
x = torch.ones(3,4)
y = torch.randn_like(x,dtype=torch.float)
x, y
```

```python
# 행렬의 덧셈
import torch
x = torch.rand(3,4)
y = torch.rand(3,4)
x, y, x+y # 덧셈을 하면, 넘파이처럼 동일한 인덱스의 요소들끼리 매치되어 합쳐진다.^그렇다면 행렬의 형식이 다르면 어떻게 될까?^
```

```python
torch.add(x,y)
```

```python
# in-place :
    # 이전에 데이터프레임을 수정하고나서, 그 데이터프레임을 다시 기존데이터 프레임에 다시 재정의를 해주는 것을
    # 알아서 해주는 기능과 동일한 기능을 텐서에서 적용하는 방법
# y = y + x    y+=x
y.add_(x)
```

```python
# 행렬 곱셈
    # 학습연산중에서 핵심적인 연산이므로 잘알아두기.
x = torch.rand(3,4)
y = torch.rand(4,5)
x, y, torch.mm(x,y), x@y # 파이썬 3.5 부터 x@y 로 동일한 기능을 제공한다.
```
