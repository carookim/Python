````python
# GPU 사용
import torch
if torch.cuda.is_available():
    device = torch.device('cuda') # gpu
    # 텐서들은 gpu에서 연산해되기 때문에 생성을 gpu를 생성
    x = torch.ones(5,device=device) # gpu
    y = torch.ones(5) # cpu
    y = y.to(device) # cpu -> gpu 로 이동
    # 변수하나는 gpu , 또다른 변수 하나는 cpu에 있다면 연산이 안된다.
    z = x+y
    print(f'{x}')
    print(f'{z.to("cpu",torch.double)}')
else:
    print('CUDA를 사용할수 없습니다.')
    device = torch.device("cpu")
````

````python
# 자동 미분 (Autograd) 매 계산과정마다 해당 계산과정의 미분값을 저장
# 파이토치 required grade = True로 설정하면 계산 그래프라는 형태로 기록 - 포토샵 되돌리기 기능
# backword() 호출하면 이과정을 거꾸로 되감으면서 각 단계의 미분값을 계산해 최종 기울기를 얻는다
````

````python
# ^
x = torch.ones(2,2, requires_grad=True)
x,  x.size()
````

````python
# ^
y = x + 2   # y도 계산 그래프에 추가
print(y, y.grad_fn)
# tensor([[3., 3.],
#         [3., 3.]], grad_fn=<AddBackward0>) <AddBackward0 object at 0x0000023543BCDAB0>
# grad_fn은 객체 이름을 통해 어떤 연산인지 간접적으로 보여줘요.
# <AddBackward0> → 덧셈(add)
# <MulBackward0> → 곱셈(multiply)
# <PowBackward0> → 거듭제곱(power)
# 즉, <AddBackward0> 자체가 “덧셈으로 만들어진 텐서”라는 표시
````

````python

z = y*y*3
out = z.mean()
z, out
````

````python
# 기울기.... 구하는 과정을... 역전파 --> 기울기 계산
# 최종결과가 out에서부터 시작해서 계산그래프를 꺼꾸로 거슬러 올라가면서 미분의 연쇄법칙을 사용해서 미분값을 계산
# out을 x에 대해서 미분 
out.backward()
x.grad
````

````python
# out = 1/4(z1+z2+z3+z4)
# z = 3y^2
# y = x + 2

# d-out/dxi = 

# 직접계산과 지원하는 자동미분법으로 나온값을 비교해보기 ^
````

````python
# 기울기 제어
# 역전파
# reaquired_grad = True연산 추적
import torch
x = torch.randn(3,requires_grad=True)
print(x)
# y는 x 로부터 연산기록을 물려받음
y = x*2

# 노름(유클리드 노름) 벡터의 크기를 측정하는 방법(원점으로 부터 벡터까지의 거리) 모든 요소들가각 제곱하고 모두 더해서 결과에 제곱근(루트)
# y의 실제 데이터의 크기를 확인해서 반복
i=0
while  y.data.norm() <1000:
    y = y*2
    i += 1
print(f'y : {y}')
print(f'반복횟수 :{i} y=x*2^i+1')
````

````python
# 벡터에 대한 역전파 (야코비안 행렬)
# 벡터를 벡터로 미분하면 결과는 행렬인 야코비안 행렬
# pytorch는 전체 야코비안 행렬을 직접 계산하는 대신 야코비안 - 벡터 곱을 계산해서 메모리를 절약
v = torch.tensor([0.1,1.0,0.001],dtype=torch.float)
    # ^ torch.tensor
y.backward(v) # y를 x에 대해 미분한 야코비안 행렬과 입력으로 제공된 벡터 v을 곱하라는 의미
print(f'x.grad : {x.grad}')
````

```
스칼라 백워드 Sclaer Backward : 역전파로 구하는 기울기가 하나 일때
out.backward()
최종점수가 숫자하나, ex) 수학공부 1시간 했을때 최종점수에 몇점의 영향을 줬는지
최종 목표가 하나(스칼라)이기 때문에 각 원인 (각 변수)의 기여도를 바로 계산

벡터 백워드 Vector Backward : 기울기가 여러개 일때
y.backward(v)
과목별 점수가 담긴 성적표(벡터) : 공부시간이 얼마나 성적에 영향을 줬는지 물어보면 애매함
                                    국어 성적인지 영어 성적인지 과학 성적인지 등등 경우의 수가 구체적으로 안추려짐
가중치 또는 중요도를 알려주는 벡터 v = torch.tensor([0.1,1.0,0.0001],dtype=torch.float)
                                                 # 영어 국어 수학 순으로 가중치 부여
중요도를 반영해서 최종적으로 공부시간의 기여도를 확인
```
