```python
import pandas as pd

# Timestamp의 배열 만들기 - 월 간격, 월의 시작일 기준
ts_ms = pd.date_range(start='2025-01-01',
                end= '2025-10-02',
                periods=100,
                # freq = 'MS',
                tz ='Asia/Seoul')

# print(ts_ms)
data = {'date' : ts_ms.strftime('%Y-%m-%d')}
df = pd.DataFrame(data)

df['date'] = pd.to_datetime(df.date) # ^
# df.date의 데이터 문자열을 **DatetimeIndex**로 변환 하여 df['date']에 넣는다.

# 년도를 추출해서 새로운 컬럼으로 넣기
df.info()
df['year'] = df.date.dt.year
df['month'] = df.date.dt.month
df['day_int'] = df.date.dt.weekday # 0 월요일

#이렇게 하면 df['date'] 컬럼만 datetime64[ns] 타입이 됩니다.
# datetime64[ns] 타입은 DatetimeIndex에서 사용하는 속성들(.year, .month, .weekday 등) 을 컬럼에서도 쓸 수 있게 해줍니다.

def convert_week(x):
    temp = ['월','화','수','목','금','토','일']
    return temp[x]
df['day_str'] = df['day_int'].apply(lambda x : convert_week(x))

# 주말(토일)은 1 나머지는 0
# 강사님 풀이
df['day_weekend'] = df['day_str'].apply(lambda x : 1 if x in ['토','일'] else 0)
df.head()
```

```python
import datetime
# 휴일 확인 함수
def check_holidays(dates, holiday_list):
    """
    dates: datetime.date 객체 리스트
    holiday_list: ItsHoliday 객체 리스트
    return: [(date, True/False, 휴일 이름 or None), ...]
    """
    results = []
    for d in dates:
        found = False
        for h in holiday_list:
            if h.date == d:
                results.append((d, True, h.name))
                found = True
                break
        if not found:
            results.append((d, False, None))
    return results

# 사용자 입력
user_input = input("확인할 날짜를 YYYY-MM-DD 형식으로 ,(콤마)로 구분해 입력하세요: ")
# 예: 2025-01-01,2025-01-02,2025-01-28

# 문자열 분리
date_strings = user_input.split(",")

# datetime.date 객체로 변환
dates_to_check = [datetime.datetime.strptime(ds.strip(), "%Y-%m-%d").date() for ds in date_strings]

results = check_holidays(dates_to_check, kr_holidays)

for r in results:
    print(r)
```
