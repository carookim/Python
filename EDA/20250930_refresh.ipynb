```python
# 여기서는 누락된 정보를 삭제하는 방법 보다는 채울 내용이 존재하기때문에,
# 그리고 누락된 데이터를 다 삭제하면 많은 데이터가 날라가기도 하므로
# dropna()가 아닌 ffill() 사용

# 누락값(NaN)을 앞 데이터로 채움 (엑셀 양식 병합 부분)
df = df.ffill()
df.head()
```

```python
# 결측치 제거 dropna()
# dropna() : 결측데이터가 포함된 행 또는 열을 제거한다.
# 행을 제거 : axis = 0 (기본값)
# 열을 제거 : axis = 1
# how='any' → 하나라도 NaN 있으면 삭제 (기본값)
# how = 'all' : 모든 값이 결측치일때 삭제
# thresh : 결측치가 아닌값의 최소개수
# 지정한 개수 이상으로 값이 채워져 있는 행/열만 남기고, 그렇지 않으면 삭제
df.dropna(thresh = 2, axis = 1) # 2개 이상의 데이터가 있는 경우에는 남긴다. 그 미만 1혹은 0의 결측데이터는 삭제한다.
                                # 해당하는 열을 제거한다.
df.dropna(how = 'all') # 모든값이 결측치일때 삭제한다.
df.dropna(thresh = 2) # 2개 이상의 데이터가 있는 경우에는 남긴다. 그 미만 1혹은 0의 결측데이터는 삭제한다.
```

```python
# 결측치 대체
# ffill() : 결측치를 직전 값으로 채운다.
# ffill = forward fill = 위쪽(이전 행)의 값을 그대로 가져와 채운다.
# 시계열 데이터, 순서가 있는 데이터에서 자주 사용
# axis : 결측치를 채울 방향, 앞에 있는 값으로 채운다. -> 그럼 반대는? bfill()
# axis=0 → 위쪽 값으로 채움 (행 기준, 기본값)
# axis=1 → 왼쪽 값으로 채움 (열 기준)
# limit : 연속된 결측치 중 몇 개까지만 채울지 지정
import numpy as np

df2 = pd.DataFrame({
    'A':[1, np.nan, np.nan, 4],
    'B':[np.nan, 2, np.nan, 3]
})

print(df2)
print(df2.ffill())                # 위쪽 값으로 채움
print(df2.ffill(limit=1))         # 연속된 NaN 중 1개만 채움 #^
print(df2.ffill(axis=1))          # 행 기준으로 왼쪽 값으로 채움

# bfill() : 결측치를 뒤쪽(다음 값) 으로 채운다.
# axis : ffill()과 반대이다. 뒤에 있는 값으로 채운다.
# axis=0 → 아래쪽 값으로 채움 (행 기준, 기본값)
# axis=1 → 오른쪽 값으로 채움 (열 기준)
# limit : 연속된 결측치 중 몇 개까지만 채울지 지정
```

```python
print(df2)
print(df2.bfill())                # 아래쪽 값으로 채움
print(df2.bfill(limit=1))         # 연속된 NaN 중 1개만 채움 #^
print(df2.bfill(axis=1))          # 행 기준으로 왼쪽 값으로 채움
```

```python
# fillna : 특정 값을 지정하여 그걸로 채운다.

df2.fillna(0) # 빈값을 0으로 채운다.
print(f'각 컬럼의 평균값 \n {df2.mean()}')
df2.fillna(df2.mean()) # 해당 컬럼의 평균 값으로 채운다.
                       # 그외 넣는 값들 정리 : .mean(), .max(), .min(), .median(),.mode()
                                            # 평균, 최대값, 최소값, 중간값, 최빈값
df2.fillna(method = 'ffill') # bfill
# DataFrame.fillna(
#     value=None,      # 단일 값, dict, Series 등으로 결측치 대체 - ()안에 그냥 입력하면 여기로 들어간다.
#     method=None,     # 'ffill' or 'bfill' : 앞/뒤 값으로 채우기
#     axis=None,       # 0=행, 1=열
#     inplace=False,   # True면 원본 변경, False면 새 DataFrame 반환
#     limit=None,      # 채우는 최대 개수
#     downcast=None    # 데이터 타입 변환 옵션
        # 데이터 타입을 임의로 변환가능하게 한 이유,
        # pandas는 컬럼 단위로 데이터 타입(dtype)을 관리하기 때문에,
        # 컬럼 안에 있는 하나의 값이 다른 타입이면 전체 컬럼의 타입이 바뀔 수 있기때문이다.
        # 예시 : 원래 int였는데 문자열이 들어오면서 object 타입이 되어버리는 경우
        #        들어가는 값의 계산으로 인해서 int였던 판다스 객체가 float가 되어버리는 경우
        # downcast=None → 아무 타입 변환 안 함
        # downcast='int' → 정수형으로 변환
        # downcast='float' → 실수형으로 변환
        # downcast='infer' → 자동으로 적절한 타입 추측
# )
```
